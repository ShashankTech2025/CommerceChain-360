```python
customers = pd.read_csv(RAW_PATH / "olist_customers_dataset.csv")

geolocation = pd.read_csv(RAW_PATH / "olist_geolocation_dataset.csv")

order_items = pd.read_csv(RAW_PATH / "olist_order_items_dataset.csv")

payments = pd.read_csv(RAW_PATH / "olist_order_payments_dataset.csv")

reviews = pd.read_csv(RAW_PATH / "olist_order_reviews_dataset.csv")

orders = pd.read_csv(RAW_PATH / "olist_orders_dataset.csv")

products = pd.read_csv(RAW_PATH / "olist_products_dataset.csv")

sellers = pd.read_csv(RAW_PATH / "olist_sellers_dataset.csv")

category_translation = pd.read_csv(
    RAW_PATH / "product_category_name_translation.csv"
)

print("All 9 datasets loaded successfully!")
```

```python
datasets = {
    "customers": customers,
    "geolocation": geolocation,
    "order_items": order_items,
    "payments": payments,
    "reviews": reviews,
    "orders": orders,
    "products": products,
    "sellers": sellers,
    "category_translation": category_translation
}

print("Total datasets:", len(datasets))
```

```python
for name, df in datasets.items():

    print("\nDATASET:", name.upper())

    display(df.describe(include='object'))
```

```python
# Task 1.8: Missing value analysis

for name, df in datasets.items():

    print("\nDATASET:", name.upper())

    missing = df.isnull().sum()

    missing_df = pd.DataFrame({
        'Missing Values': missing,
        'Missing Percentage': (missing / len(df)) * 100
    })

    display(missing_df[missing_df['Missing Values'] > 0])
```

```python
# Task 1.9: Exact duplicate analysis

duplicate_summary = []

for name, df in datasets.items():

    duplicate_count = df.duplicated().sum()

    duplicate_summary.append({
        "Dataset": name,
        "Total Rows": len(df),
        "Duplicate Rows": duplicate_count,
        "Duplicate Percentage": round(
            (duplicate_count / len(df)) * 100, 2
        )
    })

duplicate_df = pd.DataFrame(duplicate_summary)

display(duplicate_df)
```

```python
# Check important identifier columns

key_columns = {
    "customers": ["customer_id", "customer_unique_id"],
    "orders": ["order_id"],
    "order_items": ["order_id", "order_item_id"],
    "payments": ["order_id"],
    "reviews": ["review_id", "order_id"],
    "products": ["product_id"],
    "sellers": ["seller_id"]
}

for table, columns in key_columns.items():

    print("\nDATASET:", table.upper())

    for col in columns:

        total = datasets[table][col].count()
        unique = datasets[table][col].nunique()
        duplicates = total - unique

        print(
            f"{col}: Total={total}, "
            f"Unique={unique}, Repeated={duplicates}"
        )
```

```python
# Task 1.10: Relationship validation

relationships = [
    ("orders", "customer_id", "customers", "customer_id"),
    ("order_items", "order_id", "orders", "order_id"),
    ("order_items", "product_id", "products", "product_id"),
    ("order_items", "seller_id", "sellers", "seller_id"),
    ("payments", "order_id", "orders", "order_id"),
    ("reviews", "order_id", "orders", "order_id"),
]

for child, child_col, parent, parent_col in relationships:

    child_values = datasets[child][child_col].dropna()

    parent_values = set(
        datasets[parent][parent_col].dropna()
    )

    matched = child_values.isin(parent_values).sum()

    unmatched = len(child_values) - matched

    percentage = (matched / len(child_values)) * 100

    print(f"\n{child}.{child_col} -> {parent}.{parent_col}")
    print("Matched:", matched)
    print("Unmatched:", unmatched)
    print("Match Percentage:", round(percentage, 2), "%")
```

```python
# Inspect exact duplicate geolocation records

geo_duplicates = geolocation[
    geolocation.duplicated(keep=False)
]

print("Total duplicate-related records:",
      len(geo_duplicates))

display(geo_duplicates.head(10))

# Check unique postal codes
print("Unique ZIP codes:",
      geolocation[
          "geolocation_zip_code_prefix"
      ].nunique())

# Check unique coordinate combinations
print("Unique coordinate combinations:",
      geolocation[
          ["geolocation_lat", "geolocation_lng"]
      ].drop_duplicates().shape[0])
```

```python
# Investigate repeated review IDs

repeated_reviews = reviews[
    reviews.duplicated(
        subset=["review_id"],
        keep=False
    )
].sort_values("review_id")

display(repeated_reviews.head(15))

# Investigate multiple reviews associated with an order

repeated_order_reviews = reviews[
    reviews.duplicated(
        subset=["order_id"],
        keep=False
    )
].sort_values("order_id")

display(repeated_order_reviews.head(15))
```

```python
# Review ID frequency

review_id_counts = reviews["review_id"].value_counts()

print("Review IDs appearing more than once:",
      (review_id_counts > 1).sum())

display(review_id_counts[review_id_counts > 1].head(10))
```

```python
# Orders having multiple review records

reviews_per_order = reviews.groupby("order_id").size()

print("Orders with multiple reviews:",
      (reviews_per_order > 1).sum())

display(
    reviews_per_order[
        reviews_per_order > 1
    ].value_counts().sort_index()
)
```

```python
# Create data dictionary

data_dictionary = []

for table_name, df in datasets.items():

    for column in df.columns:

        data_dictionary.append({
            "Table Name": table_name,
            "Column Name": column,
            "Data Type": str(df[column].dtype),
            "Non Null Count": df[column].count(),
            "Missing Values": df[column].isnull().sum(),
            "Unique Values": df[column].nunique()
        })

data_dictionary = pd.DataFrame(data_dictionary)

display(data_dictionary)
```
