```python

# ============================================================
# COMMERCECHAIN 360 - ORDER ITEMS DATA CLEANING
# Dataset: olist_order_items_dataset.csv
# ============================================================

import pandas as pd
import numpy as np

# 1. LOAD RAW DATA
order_items = pd.read_csv(
    RAW_PATH / "olist_order_items_dataset.csv"
)

order_items_clean = order_items.copy()

print("Original Dataset Shape:", order_items_clean.shape)

# 2. STANDARDIZE COLUMN NAMES
order_items_clean.columns = (
    order_items_clean.columns.str.strip().str.lower()
)

# 3. CLEAN TEXT / IDENTIFIER COLUMNS
id_columns = ["order_id", "product_id", "seller_id"]

for col in id_columns:
    order_items_clean[col] = (
        order_items_clean[col].astype("string").str.strip()
    )

# 4. CONVERT DATA TYPES
order_items_clean["order_item_id"] = pd.to_numeric(
    order_items_clean["order_item_id"],
    errors="coerce"
).astype("Int64")

order_items_clean["price"] = pd.to_numeric(
    order_items_clean["price"],
    errors="coerce"
)

order_items_clean["freight_value"] = pd.to_numeric(
    order_items_clean["freight_value"],
    errors="coerce"
)

order_items_clean["shipping_limit_date"] = pd.to_datetime(
    order_items_clean["shipping_limit_date"],
    errors="coerce"
)

# 5. REMOVE EXACT DUPLICATE ROWS
before_duplicates = len(order_items_clean)

order_items_clean = order_items_clean.drop_duplicates()

removed_duplicates = before_duplicates - len(order_items_clean)

# 6. VALIDATE IDENTIFIERS
order_items_clean["invalid_identifier"] = (
    order_items_clean[id_columns].isna().any(axis=1) |
    order_items_clean[id_columns].eq("").any(axis=1) |
    order_items_clean["order_item_id"].isna()
)

# 7. CHECK COMPOSITE KEY DUPLICATES
duplicate_composite_keys = order_items_clean.duplicated(
    subset=["order_id", "order_item_id"],
    keep=False
)

# Preserve these records for investigation.
order_items_clean["duplicate_composite_key"] = (
    duplicate_composite_keys
)

# 8. FLAG INVALID NUMERIC VALUES
order_items_clean["invalid_price"] = (
    order_items_clean["price"].isna() |
    (order_items_clean["price"] <= 0)
)

order_items_clean["invalid_freight"] = (
    order_items_clean["freight_value"].isna() |
    (order_items_clean["freight_value"] < 0)
)

# 9. FLAG MISSING OR INVALID SHIPPING DATES
order_items_clean["invalid_shipping_date"] = (
    order_items_clean["shipping_limit_date"].isna()
)

# 10. CREATE TOTAL ITEM VALUE
order_items_clean["total_item_value"] = (
    order_items_clean["price"] +
    order_items_clean["freight_value"]
)

# 11. CREATE DATA QUALITY STATUS
quality_issues = (
    order_items_clean["invalid_identifier"] |
    order_items_clean["duplicate_composite_key"] |
    order_items_clean["invalid_price"] |
    order_items_clean["invalid_freight"] |
    order_items_clean["invalid_shipping_date"]
)

order_items_clean["data_quality_status"] = np.where(
    quality_issues,
    "Review Required",
    "Valid"
)

# 12. FINAL DATA QUALITY REPORT
print("\n========== CLEANING SUMMARY ==========")

print("Original Rows:", len(order_items))
print("Rows After Exact Duplicate Removal:",
      len(order_items_clean))
print("Exact Duplicates Removed:", removed_duplicates)

print("\nMissing Values:")
print(order_items_clean.isnull().sum())

print("\nInvalid Identifiers:",
      order_items_clean["invalid_identifier"].sum())

print("Duplicate Composite Keys:",
      order_items_clean["duplicate_composite_key"].sum())

print("Invalid Prices:",
      order_items_clean["invalid_price"].sum())

print("Invalid Freight Values:",
      order_items_clean["invalid_freight"].sum())

print("Invalid Shipping Dates:",
      order_items_clean["invalid_shipping_date"].sum())

print("\nData Quality Status:")
print(order_items_clean["data_quality_status"].value_counts())

# 13. EXPORT CLEANED DATASET
items_file = PROCESSED_PATH / "order_items_clean.csv"

order_items_clean.to_csv(items_file, index=False)

# 14. VERIFY EXPORTED DATASET
items_verified = pd.read_csv(
    items_file,
    dtype={
        "order_id": "string",
        "product_id": "string",
        "seller_id": "string"
    },
    parse_dates=["shipping_limit_date"]
)

print("\n========== EXPORT VERIFICATION ==========")
print("Exported Shape:", items_verified.shape)

assert items_verified.shape == order_items_clean.shape

print("Order Items Dataset Exported Successfully!")
print("Saved Location:", items_file)
```

```python

# ============================================================
# COMMERCECHAIN 360 - PAYMENT DATA CLEANING
# ============================================================

import pandas as pd
import numpy as np

# 1. LOAD DATA
payments = pd.read_csv(
    RAW_PATH / "olist_order_payments_dataset.csv"
)

payments_clean = payments.copy()

print("Original Shape:", payments_clean.shape)

# 2. STANDARDIZE COLUMN NAMES
payments_clean.columns = (
    payments_clean.columns.str.strip().str.lower()
)

# 3. CLEAN IDENTIFIER
payments_clean["order_id"] = (
    payments_clean["order_id"].astype("string").str.strip()
)

# 4. CONVERT DATA TYPES
payments_clean["payment_sequential"] = pd.to_numeric(
    payments_clean["payment_sequential"],
    errors="coerce"
).astype("Int64")

payments_clean["payment_installments"] = pd.to_numeric(
    payments_clean["payment_installments"],
    errors="coerce"
).astype("Int64")

payments_clean["payment_value"] = pd.to_numeric(
    payments_clean["payment_value"],
    errors="coerce"
)

payments_clean["payment_type"] = (
    payments_clean["payment_type"]
    .astype("string")
    .str.strip()
    .str.lower()
)

# 5. REMOVE EXACT DUPLICATES
before = len(payments_clean)

payments_clean = payments_clean.drop_duplicates()

removed_duplicates = before - len(payments_clean)

# 6. VALIDATE IDENTIFIERS
payments_clean["invalid_identifier"] = (
    payments_clean["order_id"].isna() |
    payments_clean["order_id"].eq("") |
    payments_clean["payment_sequential"].isna() |
    (payments_clean["payment_sequential"] <= 0)
)

# 7. CHECK COMPOSITE KEY
payments_clean["duplicate_composite_key"] = (
    payments_clean.duplicated(
        subset=["order_id", "payment_sequential"],
        keep=False
    )
)

# 8. VALIDATE PAYMENT VALUES
payments_clean["invalid_payment_value"] = (
    payments_clean["payment_value"].isna() |
    (payments_clean["payment_value"] < 0)
)

# Zero payment values are flagged separately for investigation
payments_clean["zero_payment_value"] = (
    payments_clean["payment_value"] == 0
)

# 9. VALIDATE INSTALLMENTS
payments_clean["invalid_installments"] = (
    payments_clean["payment_installments"].isna() |
    (payments_clean["payment_installments"] <= 0)
)

# 10. VALIDATE PAYMENT TYPE
valid_payment_types = [
    "credit_card",
    "debit_card",
    "boleto",
    "voucher",
    "not_defined"
]

payments_clean["invalid_payment_type"] = (
    payments_clean["payment_type"].isna() |
    ~payments_clean["payment_type"].isin(valid_payment_types)
)

# 11. CREATE QUALITY STATUS
quality_issues = (
    payments_clean["invalid_identifier"] |
    payments_clean["duplicate_composite_key"] |
    payments_clean["invalid_payment_value"] |
    payments_clean["invalid_installments"] |
    payments_clean["invalid_payment_type"]
)

payments_clean["data_quality_status"] = np.where(
    quality_issues,
    "Review Required",
    "Valid"
)

# 12. CLEANING SUMMARY
print("\n========== CLEANING SUMMARY ==========")

print("Original Rows:", len(payments))
print("Final Rows:", len(payments_clean))
print("Exact Duplicates Removed:", removed_duplicates)

print("\nMissing Values:")
print(payments_clean.isnull().sum())

print("\nInvalid Identifiers:",
      payments_clean["invalid_identifier"].sum())

print("Duplicate Composite Keys:",
      payments_clean["duplicate_composite_key"].sum())

print("Invalid Payment Values:",
      payments_clean["invalid_payment_value"].sum())

print("Zero Payment Values:",
      payments_clean["zero_payment_value"].sum())

print("Invalid Installments:",
      payments_clean["invalid_installments"].sum())

print("Invalid Payment Types:",
      payments_clean["invalid_payment_type"].sum())

print("\nPayment Type Distribution:")
print(payments_clean["payment_type"].value_counts(dropna=False))

print("\nData Quality Status:")
print(payments_clean["data_quality_status"].value_counts())

# 13. EXPORT
payment_file = PROCESSED_PATH / "payments_clean.csv"

payments_clean.to_csv(payment_file, index=False)

# 14. VERIFY
payments_verified = pd.read_csv(
    payment_file,
    dtype={"order_id": "string"}
)

print("\n========== EXPORT VERIFICATION ==========")

print("Verified Shape:", payments_verified.shape)

assert payments_verified.shape == payments_clean.shape

print("Payment Dataset Exported Successfully!")
print("Saved Location:", payment_file)
```

```python

# ============================================================
# COMMERCECHAIN 360 - ORDER REVIEWS DATA CLEANING
# ============================================================

import pandas as pd
import numpy as np

# 1. LOAD DATA
reviews = pd.read_csv(
    RAW_PATH / "olist_order_reviews_dataset.csv"
)

reviews_clean = reviews.copy()

print("Original Shape:", reviews_clean.shape)

# 2. STANDARDIZE COLUMN NAMES
reviews_clean.columns = (
    reviews_clean.columns.str.strip().str.lower()
)

# 3. CLEAN IDENTIFIERS
for col in ["review_id", "order_id"]:
    reviews_clean[col] = (
        reviews_clean[col].astype("string").str.strip()
    )

# 4. CONVERT REVIEW SCORE
reviews_clean["review_score"] = pd.to_numeric(
    reviews_clean["review_score"],
    errors="coerce"
).astype("Int64")

# 5. CONVERT DATE COLUMNS
date_columns = [
    "review_creation_date",
    "review_answer_timestamp"
]

for col in date_columns:
    reviews_clean[col] = pd.to_datetime(
        reviews_clean[col],
        errors="coerce"
    )

# 6. STANDARDIZE TEXT COLUMNS
text_columns = [
    "review_comment_title",
    "review_comment_message"
]

for col in text_columns:
    reviews_clean[col] = (
        reviews_clean[col]
        .astype("string")
        .str.strip()
    )

    # Convert empty strings to missing values
    reviews_clean[col] = reviews_clean[col].replace("", pd.NA)

# 7. REMOVE EXACT DUPLICATES
before = len(reviews_clean)

reviews_clean = reviews_clean.drop_duplicates()

removed_duplicates = before - len(reviews_clean)

# 8. VALIDATE IDENTIFIERS
reviews_clean["invalid_identifier"] = (
    reviews_clean["review_id"].isna() |
    reviews_clean["review_id"].eq("") |
    reviews_clean["order_id"].isna() |
    reviews_clean["order_id"].eq("")
)

# 9. VALIDATE REVIEW SCORES
reviews_clean["invalid_review_score"] = (
    reviews_clean["review_score"].isna() |
    ~reviews_clean["review_score"].between(1, 5)
)

# 10. FLAG DUPLICATE REVIEW IDs
reviews_clean["duplicate_review_id"] = (
    reviews_clean.duplicated(
        subset=["review_id"],
        keep=False
    )
)

# 11. FLAG MULTIPLE REVIEWS PER ORDER
reviews_clean["multiple_reviews_per_order"] = (
    reviews_clean.duplicated(
        subset=["order_id"],
        keep=False
    )
)

# 12. FLAG INVALID DATES
reviews_clean["invalid_review_creation_date"] = (
    reviews_clean["review_creation_date"].isna()
)

reviews_clean["invalid_review_answer_date"] = (
    reviews_clean["review_answer_timestamp"].isna()
)

# 13. FLAG DATE ORDER ISSUES
reviews_clean["answer_before_creation"] = (
    reviews_clean["review_creation_date"].notna() &
    reviews_clean["review_answer_timestamp"].notna() &
    (
        reviews_clean["review_answer_timestamp"] <
        reviews_clean["review_creation_date"]
    )
)

# 14. CREATE REVIEW LENGTH FEATURES
reviews_clean["review_message_length"] = (
    reviews_clean["review_comment_message"]
    .str.len()
)

reviews_clean["review_title_length"] = (
    reviews_clean["review_comment_title"]
    .str.len()
)

# 15. CREATE QUALITY STATUS
quality_issues = (
    reviews_clean["invalid_identifier"] |
    reviews_clean["invalid_review_score"] |
    reviews_clean["duplicate_review_id"] |
    reviews_clean["invalid_review_creation_date"] |
    reviews_clean["answer_before_creation"]
)

reviews_clean["data_quality_status"] = np.where(
    quality_issues,
    "Review Required",
    "Valid"
)

# 16. CLEANING SUMMARY
print("\n========== CLEANING SUMMARY ==========")

print("Original Rows:", len(reviews))
print("Final Rows:", len(reviews_clean))
print("Exact Duplicates Removed:", removed_duplicates)

print("\nMissing Values:")
print(reviews_clean.isnull().sum())

print("\nInvalid Identifiers:",
      reviews_clean["invalid_identifier"].sum())

print("Invalid Review Scores:",
      reviews_clean["invalid_review_score"].sum())

print("Duplicate Review IDs:",
      reviews_clean["duplicate_review_id"].sum())

print("Multiple Reviews Per Order:",
      reviews_clean["multiple_reviews_per_order"].sum())

print("Invalid Creation Dates:",
      reviews_clean["invalid_review_creation_date"].sum())

print("Invalid Answer Dates:",
      reviews_clean["invalid_review_answer_date"].sum())

print("Answer Before Creation:",
      reviews_clean["answer_before_creation"].sum())

print("\nReview Score Distribution:")
print(reviews_clean["review_score"].value_counts().sort_index())

print("\nData Quality Status:")
print(reviews_clean["data_quality_status"].value_counts())

# 17. EXPORT
review_file = PROCESSED_PATH / "reviews_clean.csv"

reviews_clean.to_csv(review_file, index=False)

# 18. VERIFY
reviews_verified = pd.read_csv(
    review_file,
    dtype={
        "review_id": "string",
        "order_id": "string"
    },
    parse_dates=[
        "review_creation_date",
        "review_answer_timestamp"
    ]
)

print("\n========== EXPORT VERIFICATION ==========")

print("Verified Shape:", reviews_verified.shape)

assert reviews_verified.shape == reviews_clean.shape

print("Reviews Dataset Exported Successfully!")
print("Saved Location:", review_file)
```

```python

# ============================================================
# COMMERCECHAIN 360 - ORDERS DATA CLEANING
# ============================================================

import pandas as pd
import numpy as np

# 1. LOAD DATA
orders = pd.read_csv(
    RAW_PATH / "olist_orders_dataset.csv"
)

orders_clean = orders.copy()

print("Original Shape:", orders_clean.shape)

# 2. STANDARDIZE COLUMN NAMES
orders_clean.columns = (
    orders_clean.columns.str.strip().str.lower()
)

# 3. CLEAN IDENTIFIERS
for col in ["order_id", "customer_id"]:
    orders_clean[col] = (
        orders_clean[col].astype("string").str.strip()
    )

# 4. CONVERT DATE COLUMNS
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    orders_clean[col] = pd.to_datetime(
        orders_clean[col],
        errors="coerce"
    )

# 5. STANDARDIZE ORDER STATUS
orders_clean["order_status"] = (
    orders_clean["order_status"]
    .astype("string")
    .str.strip()
    .str.lower()
)

# 6. REMOVE EXACT DUPLICATES
before = len(orders_clean)

orders_clean = orders_clean.drop_duplicates()

removed_duplicates = before - len(orders_clean)

# 7. VALIDATE IDENTIFIERS
orders_clean["invalid_identifier"] = (
    orders_clean["order_id"].isna() |
    orders_clean["order_id"].eq("") |
    orders_clean["customer_id"].isna() |
    orders_clean["customer_id"].eq("")
)

# 8. VALIDATE ORDER STATUS
valid_statuses = [
    "created",
    "approved",
    "invoiced",
    "processing",
    "shipped",
    "delivered",
    "unavailable",
    "canceled"
]

orders_clean["invalid_order_status"] = (
    orders_clean["order_status"].isna() |
    ~orders_clean["order_status"].isin(valid_statuses)
)

# 9. VALIDATE ORDER PURCHASE DATE
orders_clean["invalid_purchase_date"] = (
    orders_clean["order_purchase_timestamp"].isna()
)

# 10. CHECK DATE SEQUENCE
orders_clean["approval_before_purchase"] = (
    orders_clean["order_approved_at"].notna() &
    (
        orders_clean["order_approved_at"] <
        orders_clean["order_purchase_timestamp"]
    )
)

orders_clean["carrier_before_approval"] = (
    orders_clean["order_delivered_carrier_date"].notna() &
    orders_clean["order_approved_at"].notna() &
    (
        orders_clean["order_delivered_carrier_date"] <
        orders_clean["order_approved_at"]
    )
)

orders_clean["delivery_before_purchase"] = (
    orders_clean["order_delivered_customer_date"].notna() &
    (
        orders_clean["order_delivered_customer_date"] <
        orders_clean["order_purchase_timestamp"]
    )
)

# 11. CREATE DELIVERY TIME FEATURES
orders_clean["delivery_days"] = (
    orders_clean["order_delivered_customer_date"] -
    orders_clean["order_purchase_timestamp"]
).dt.total_seconds() / 86400

orders_clean["estimated_delivery_days"] = (
    orders_clean["order_estimated_delivery_date"] -
    orders_clean["order_purchase_timestamp"]
).dt.total_seconds() / 86400

# 12. CALCULATE DELIVERY DELAY
orders_clean["delivery_delay_days"] = (
    orders_clean["order_delivered_customer_date"] -
    orders_clean["order_estimated_delivery_date"]
).dt.total_seconds() / 86400

# 13. CREATE DELIVERY PERFORMANCE STATUS
orders_clean["delivery_performance"] = np.select(
    [
        orders_clean["order_status"].eq("delivered") &
        orders_clean["delivery_delay_days"].notna() &
        (orders_clean["delivery_delay_days"] <= 0),

        orders_clean["order_status"].eq("delivered") &
        orders_clean["delivery_delay_days"].notna() &
        (orders_clean["delivery_delay_days"] > 0),

        orders_clean["order_status"].eq("delivered")
    ],
    [
        "On Time or Early",
        "Late Delivery",
        "Delivery Date Unavailable"
    ],
    default="Not Delivered"
)

# 14. CREATE DATA QUALITY STATUS
quality_issues = (
    orders_clean["invalid_identifier"] |
    orders_clean["invalid_order_status"] |
    orders_clean["invalid_purchase_date"] |
    orders_clean["approval_before_purchase"] |
    orders_clean["carrier_before_approval"] |
    orders_clean["delivery_before_purchase"]
)

orders_clean["data_quality_status"] = np.where(
    quality_issues,
    "Review Required",
    "Valid"
)

# 15. CLEANING SUMMARY
print("\n========== CLEANING SUMMARY ==========")

print("Original Rows:", len(orders))
print("Final Rows:", len(orders_clean))
print("Exact Duplicates Removed:", removed_duplicates)

print("\nMissing Values:")
print(orders_clean.isnull().sum())

print("\nInvalid Identifiers:",
      orders_clean["invalid_identifier"].sum())

print("Invalid Order Status:",
      orders_clean["invalid_order_status"].sum())

print("Invalid Purchase Dates:",
      orders_clean["invalid_purchase_date"].sum())

print("Approval Before Purchase:",
      orders_clean["approval_before_purchase"].sum())

print("Carrier Before Approval:",
      orders_clean["carrier_before_approval"].sum())

print("Delivery Before Purchase:",
      orders_clean["delivery_before_purchase"].sum())

print("\nOrder Status Distribution:")
print(orders_clean["order_status"].value_counts())

print("\nDelivery Performance:")
print(orders_clean["delivery_performance"].value_counts())

print("\nData Quality Status:")
print(orders_clean["data_quality_status"].value_counts())

# 16. EXPORT CLEANED DATASET
orders_file = PROCESSED_PATH / "orders_clean.csv"

orders_clean.to_csv(orders_file, index=False)

# 17. VERIFY EXPORTED FILE
orders_verified = pd.read_csv(
    orders_file,
    dtype={
        "order_id": "string",
        "customer_id": "string"
    },
    parse_dates=date_columns
)

print("\n========== EXPORT VERIFICATION ==========")

print("Verified Shape:", orders_verified.shape)

assert orders_verified.shape == orders_clean.shape

print("Orders Dataset Exported Successfully!")
print("Saved Location:", orders_file)
```

```python

# ============================================================
# COMMERCECHAIN 360 - PRODUCTS DATA CLEANING
# ============================================================

import pandas as pd
import numpy as np

# 1. LOAD DATA
products = pd.read_csv(
    RAW_PATH / "olist_products_dataset.csv"
)

products_clean = products.copy()

print("Original Shape:", products_clean.shape)

# 2. STANDARDIZE COLUMN NAMES
products_clean.columns = (
    products_clean.columns.str.strip().str.lower()
)

# 3. CLEAN PRODUCT IDENTIFIER
products_clean["product_id"] = (
    products_clean["product_id"].astype("string").str.strip()
)

# 4. STANDARDIZE PRODUCT CATEGORY
products_clean["product_category_name"] = (
    products_clean["product_category_name"]
    .astype("string")
    .str.strip()
    .str.lower()
)

products_clean["product_category_name"] = (
    products_clean["product_category_name"].replace("", pd.NA)
)

# 5. CONVERT NUMERIC COLUMNS
numeric_columns = [
    "product_name_lenght",
    "product_description_lenght",
    "product_photos_qty",
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]

for col in numeric_columns:
    products_clean[col] = pd.to_numeric(
        products_clean[col],
        errors="coerce"
    )

# 6. REMOVE EXACT DUPLICATES
before = len(products_clean)

products_clean = products_clean.drop_duplicates()

removed_duplicates = before - len(products_clean)

# 7. VALIDATE PRODUCT IDENTIFIER
products_clean["invalid_product_id"] = (
    products_clean["product_id"].isna() |
    products_clean["product_id"].eq("")
)

products_clean["duplicate_product_id"] = (
    products_clean.duplicated(
        subset=["product_id"],
        keep=False
    )
)

# 8. VALIDATE PRODUCT DIMENSIONS
dimension_columns = [
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]

for col in dimension_columns:
    products_clean[f"invalid_{col}"] = (
        products_clean[col].isna() |
        (products_clean[col] <= 0)
    )

# 9. VALIDATE OTHER PRODUCT ATTRIBUTES
products_clean["invalid_photos_qty"] = (
    products_clean["product_photos_qty"].isna() |
    (products_clean["product_photos_qty"] < 0)
)

products_clean["invalid_name_length"] = (
    products_clean["product_name_lenght"].isna() |
    (products_clean["product_name_lenght"] < 0)
)

products_clean["invalid_description_length"] = (
    products_clean["product_description_lenght"].isna() |
    (products_clean["product_description_lenght"] < 0)
)

# 10. CREATE PRODUCT VOLUME
products_clean["product_volume_cm3"] = (
    products_clean["product_length_cm"] *
    products_clean["product_height_cm"] *
    products_clean["product_width_cm"]
)

# 11. CREATE DIMENSIONAL WEIGHT
# Standard carrier volumetric divisor: 6000.
# This is an analytical estimate, not an Olist-provided field.
products_clean["dimensional_weight_kg"] = (
    products_clean["product_volume_cm3"] / 6000
)

# 12. CREATE QUALITY STATUS
quality_columns = [
    "invalid_product_id",
    "duplicate_product_id",
    "invalid_product_weight_g",
    "invalid_product_length_cm",
    "invalid_product_height_cm",
    "invalid_product_width_cm",
    "invalid_photos_qty",
    "invalid_name_length",
    "invalid_description_length"
]

quality_issues = products_clean[quality_columns].any(axis=1)

products_clean["data_quality_status"] = np.where(
    quality_issues,
    "Review Required",
    "Valid"
)

# 13. CLEANING SUMMARY
print("\n========== CLEANING SUMMARY ==========")

print("Original Rows:", len(products))
print("Final Rows:", len(products_clean))
print("Exact Duplicates Removed:", removed_duplicates)

print("\nMissing Values:")
print(products_clean.isnull().sum())

print("\nInvalid Product IDs:",
      products_clean["invalid_product_id"].sum())

print("Duplicate Product IDs:",
      products_clean["duplicate_product_id"].sum())

print("\nInvalid Dimensions:")
for col in dimension_columns:
    print(col, ":", products_clean[f"invalid_{col}"].sum())

print("\nInvalid Photo Quantities:",
      products_clean["invalid_photos_qty"].sum())

print("Invalid Product Name Lengths:",
      products_clean["invalid_name_length"].sum())

print("Invalid Description Lengths:",
      products_clean["invalid_description_length"].sum())

print("\nTop 10 Product Categories:")
print(
    products_clean["product_category_name"]
    .value_counts()
    .head(10)
)

print("\nData Quality Status:")
print(products_clean["data_quality_status"].value_counts())

# 14. EXPORT CLEANED DATASET
products_file = PROCESSED_PATH / "products_clean.csv"

products_clean.to_csv(products_file, index=False)

# 15. VERIFY EXPORT
products_verified = pd.read_csv(
    products_file,
    dtype={"product_id": "string"}
)

print("\n========== EXPORT VERIFICATION ==========")

print("Verified Shape:", products_verified.shape)

assert products_verified.shape == products_clean.shape

print("Products Dataset Exported Successfully!")
print("Saved Location:", products_file)
```

```python

# ============================================================
# COMMERCECHAIN 360 - FINAL INTEGRATION VALIDATION
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

# 1. LOAD ALL CLEANED DATASETS

cleaned_files = {
    "customers": "customers_clean.csv",
    "geolocation": "geolocation_clean.csv",
    "order_items": "order_items_clean.csv",
    "payments": "payments_clean.csv",
    "reviews": "reviews_clean.csv",
    "orders": "orders_clean.csv",
    "products": "products_clean.csv",
    "sellers": "sellers_clean.csv",
    "category_translation": "category_translation_clean.csv"
}

cleaned_datasets = {}

for name, filename in cleaned_files.items():
    file_path = PROCESSED_PATH / filename

    if not file_path.exists():
        raise FileNotFoundError(f"Missing file: {file_path}")

    cleaned_datasets[name] = pd.read_csv(
        file_path,
        dtype={
            "customer_id": "string",
            "customer_unique_id": "string",
            "order_id": "string",
            "product_id": "string",
            "seller_id": "string",
            "review_id": "string",
            "customer_zip_code_prefix": "string",
            "seller_zip_code_prefix": "string",
            "geolocation_zip_code_prefix": "string"
        }
    )

print("All nine cleaned datasets loaded successfully!")

# 2. DATASET INVENTORY

inventory = []

for name, df in cleaned_datasets.items():
    inventory.append({
        "Dataset": name,
        "Rows": len(df),
        "Columns": len(df.columns),
        "Duplicate Rows": df.duplicated().sum(),
        "Missing Cells": df.isnull().sum().sum()
    })

inventory_df = pd.DataFrame(inventory)

print("\n========== DATASET INVENTORY ==========")
display(inventory_df)

# 3. VALIDATE PRIMARY KEYS

primary_keys = {
    "customers": "customer_id",
    "orders": "order_id",
    "products": "product_id",
    "sellers": "seller_id",
    "category_translation": "product_category_name"
}

print("\n========== PRIMARY KEY VALIDATION ==========")

for dataset, key in primary_keys.items():
    df = cleaned_datasets[dataset]

    print(
        dataset,
        "| Null keys:", df[key].isna().sum(),
        "| Duplicate keys:", df[key].duplicated().sum()
    )

# 4. VALIDATE FOREIGN KEY RELATIONSHIPS

customers_df = cleaned_datasets["customers"]
orders_df = cleaned_datasets["orders"]
items_df = cleaned_datasets["order_items"]
payments_df = cleaned_datasets["payments"]
reviews_df = cleaned_datasets["reviews"]
products_df = cleaned_datasets["products"]
sellers_df = cleaned_datasets["sellers"]

relationships = [
    (
        "orders.customer_id",
        orders_df["customer_id"],
        customers_df["customer_id"]
    ),
    (
        "order_items.order_id",
        items_df["order_id"],
        orders_df["order_id"]
    ),
    (
        "order_items.product_id",
        items_df["product_id"],
        products_df["product_id"]
    ),
    (
        "order_items.seller_id",
        items_df["seller_id"],
        sellers_df["seller_id"]
    ),
    (
        "payments.order_id",
        payments_df["order_id"],
        orders_df["order_id"]
    ),
    (
        "reviews.order_id",
        reviews_df["order_id"],
        orders_df["order_id"]
    )
]

relationship_results = []

print("\n========== RELATIONSHIP VALIDATION ==========")

for name, child, parent in relationships:

    child_values = child.dropna()
    parent_values = set(parent.dropna())

    unmatched = (~child_values.isin(parent_values)).sum()

    match_rate = (
        (len(child_values) - unmatched) / len(child_values) * 100
        if len(child_values) else 0
    )

    relationship_results.append({
        "Relationship": name,
        "Records Checked": len(child_values),
        "Unmatched Records": unmatched,
        "Match Percentage": round(match_rate, 2)
    })

relationship_df = pd.DataFrame(relationship_results)

display(relationship_df)

# 5. PRODUCT CATEGORY TRANSLATION COVERAGE

translation_df = cleaned_datasets["category_translation"]

category_check = products_df[
    ["product_category_name"]
].merge(
    translation_df[
        ["product_category_name", "product_category_name_english"]
    ],
    on="product_category_name",
    how="left"
)

print("\n========== CATEGORY TRANSLATION ==========")

print(
    "Products with untranslated categories:",
    category_check[
        category_check["product_category_name"].notna() &
        category_check["product_category_name_english"].isna()
    ].shape[0]
)

print(
    "Products with missing category:",
    products_df["product_category_name"].isna().sum()
)

# 6. SAVE VALIDATION REPORTS

REPORT_PATH = PROJECT_PATH / "reports"
REPORT_PATH.mkdir(parents=True, exist_ok=True)

inventory_df.to_csv(
    REPORT_PATH / "final_dataset_inventory.csv",
    index=False
)

relationship_df.to_csv(
    REPORT_PATH / "final_relationship_validation.csv",
    index=False
)

inventory_df.to_excel(
    REPORT_PATH / "CommerceChain_Final_Data_Inventory.xlsx",
    index=False
)

relationship_df.to_excel(
    REPORT_PATH / "CommerceChain_Relationship_Validation.xlsx",
    index=False
)

print("\n========== FINAL VALIDATION ==========")
print("Dataset inventory exported!")
print("Relationship report exported!")
print("Reports saved in:", REPORT_PATH)
```
